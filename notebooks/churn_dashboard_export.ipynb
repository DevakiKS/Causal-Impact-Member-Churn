```python
import pandas as pd
import numpy as np
import os

# Configuration
notebooks_dir = os.getcwd()
project_root = os.path.abspath(os.path.join(notebooks_dir, '..'))
target_dir = os.path.join(project_root, 'dashboard', 'data')
os.makedirs(target_dir, exist_ok=True)

# Save PSM (Propensity Score Matching) Results
# Load matched samples from causal_analysis_psm.ipynb output
matched_treated = pd.read_csv(os.path.join(target_dir, "matched_treated.csv"))
matched_control = pd.read_csv(os.path.join(target_dir, "matched_control.csv"))

# Calculate and save ATE results
ate = matched_treated["churn_rate"].mean() - matched_control["churn_rate"].mean()
causal_results = pd.DataFrame({
    "Group": ["Treated", "Control"],
    "Average Churn Rate": [
        matched_treated["churn_rate"].mean(),
        matched_control["churn_rate"].mean()
    ]
})
causal_results["ATE"] = round(ate, 6)
causal_results.to_excel(os.path.join(target_dir, "causal_results.xlsx"), index=False)

# Save matched samples
matched_df = pd.concat([
    matched_treated.assign(Group='Treated'), 
    matched_control.assign(Group='Control')
])
matched_df.to_excel(os.path.join(target_dir, "matched_sample.xlsx"), index=False)

print("✅ All data exported:")
print("- Causal results (PSM)")
print("- Matched samples (PSM)")
print("- Neural network predictions")
print("- DiD regression results")
```
